```python
import pandas as pd
```

```python
def fmt(x):
    return f"{x:.2f}" if isinstance(x, float) else x
```

# No PPK

```python
ppk_false = pd.read_csv("./gesses_ppk-False.csv")
ppk_false
```

```python
best_ppk_false = ppk_false.sort_values(by="sigmoid_steady_state prediction", ascending=False).iloc[0]
best_ppk_false.index.name = "Property"
best_ppk_false.name = "Value"
```

```python
print(best_ppk_false.map(fmt).to_markdown())
```
