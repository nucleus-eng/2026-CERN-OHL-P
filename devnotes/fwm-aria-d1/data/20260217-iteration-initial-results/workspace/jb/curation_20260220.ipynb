```python
## Convert the "Read" column to a set of columns that are one hot for each read type
real_one_hot = pd.get_dummies(df["Read"].iloc[1:], prefix="Read").astype(int) ## THIS HAD iloc[2:], BUT SHOULD HAVE iloc[1:]
real_one_hot.loc[0] = "feature"
df = pd.concat([df, real_one_hot], axis=1).drop("Read", axis=1)
```

```python
## Convert categorical "Product" to one hot (which is just plamGFP or not (= deGFP))
product_bool = (df["Product"].iloc[1:] == "plamGFP").astype(int)
product_bool.loc[0] = "feature"
df["Product_plamGFP"] = product_bool
df = df.drop("Product", axis=1)
```

```python
# Drop columns that have only constant values
df = df.drop(df.columns[df[1:].nunique() == 1], axis=1)
```
