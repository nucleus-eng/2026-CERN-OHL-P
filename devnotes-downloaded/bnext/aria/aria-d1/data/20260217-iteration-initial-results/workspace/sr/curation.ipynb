```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("round_0_failures_w_headers.csv")
df.loc[0, "experiment_name"] = "id"
```

```python
df.iloc[0:2].T.value_counts(dropna=False)
```

```python
df.loc[:, df.iloc[1] == "discrete"].iloc[2:, :].value_counts()
```

```python
real_one_hot = pd.get_dummies(df["Read"].iloc[2:], prefix="Read").astype(int)
real_one_hot.loc[0] = "feature"
df = pd.concat([df, real_one_hot], axis=1).drop("Read", axis=1)
```

```python
product_bool = (df["Product"].iloc[2:] == "plamGFP").astype(int)
product_bool.loc[0] = "feature"
df["Product_plamGFP"] = product_bool
df = df.drop("Product", axis=1)
```

```python
df = df.drop(df.columns[df[2:].nunique() == 1], axis=1)
```

```python
corr = df.drop(["experiment_id", "experiment_name"], axis=1).iloc[2:, :].corr()
plt.imshow(np.abs(corr) > .99)
plt.show()
corr
```
